```python
# 654. 最大二叉树
# Definition for a binary tree node.
# class TreeNode(object):
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution(object):
    def constructMaximumBinaryTree(self, nums):
        """
        :type nums: List[int]
        :rtype: Optional[TreeNode]
        """
        if len(nums) == 0:
            return None
        val = max(nums)
        maxIndex = nums.index(val)
        cur = TreeNode(val)
        numsLeft = nums[:maxIndex]
        numsRight = nums[maxIndex + 1:]
        cur.left = self.constructMaximumBinaryTree(numsLeft)
        cur.right = self.constructMaximumBinaryTree(numsRight)
        return cur


        
```

```python
# 617.合并二叉树
# Definition for a binary tree node.
# class TreeNode(object):
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution(object):
    def mergeTree(self, cur1, cur2):
        if cur1 == None and cur2 == None:
            return None
        if cur1 == None and cur2 != None:
            return cur2
        if cur1 != None and cur2 == None:
            return cur1
        cur = TreeNode(cur1.val + cur2.val)
        cur.left = self.mergeTree(cur1.left, cur2.left)
        cur.right = self.mergeTree(cur1.right, cur2.right)
        return cur
    def mergeTrees(self, root1, root2):
        """
        :type root1: Optional[TreeNode]
        :type root2: Optional[TreeNode]
        :rtype: Optional[TreeNode]
        """
        return self.mergeTree(root1, root2)
```

```python
# 700. 二叉搜索树中的搜索（递归）
# Definition for a binary tree node.
# class TreeNode(object):
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution(object):
    def searchBST(self, root, val):
        """
        :type root: Optional[TreeNode]
        :type val: int
        :rtype: Optional[TreeNode]
        """
        if root == None:
            return None
        if root.val == val:
            return root
        if root.val < val:
            return self.searchBST(root.right, val)
        if root.val > val:
            return self.searchBST(root.left, val)
        
```

```python
# 700. 二叉搜索树中的搜索（迭代）
# Definition for a binary tree node.
# class TreeNode(object):
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution(object):
    def searchBST(self, root, val):
        """
        :type root: Optional[TreeNode]
        :type val: int
        :rtype: Optional[TreeNode]
        """
        cur = root
        while cur:
            if cur.val == val:
                return cur
            elif cur.val > val:
                cur = cur.left
            else:
                cur = cur.right
        return None
```

```python
# 98. 验证二叉搜索树(二叉搜索树的中序遍历有序，用数组存储)
# Definition for a binary tree node.
# class TreeNode(object):
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution(object):
    def __init__(self):
        self.nodeList = []
    def inorderTree(self, root):
        if root == None:
            return
        self.inorderTree(root.left)
        self.nodeList.append(root.val)
        self.inorderTree(root.right)
        return
    def isValidBST(self, root):
        """
        :type root: Optional[TreeNode]
        :rtype: bool
        """
        self.inorderTree(root)
        # print(self.nodeList)
        for i in range(len(self.nodeList) - 1):
            if self.nodeList[i] >= self.nodeList[i + 1]:
                return False
        return True

        
```

```python
# 98. 验证二叉搜索树(二叉搜索树的中序遍历有序，数记录)
# Definition for a binary tree node.
# class TreeNode(object):
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution(object):
    def __init__(self):
        self.maxVal = float("-inf")
    def isValidBST(self, root):
        """
        :type root: Optional[TreeNode]
        :rtype: bool
        """
        if root == None:
            return True
        left = self.isValidBST(root.left)
        if  self.maxVal >= root.val:
            return False
        else:
            self.maxVal = root.val
        right = self.isValidBST(root.right)

        return left and right
        
```
