```python
# 77. 组合
class Solution(object):
    def __init__(self):
        self.result = []
    def backTracking(self, startIndex, n, k, path):
        if len(path) == k:
            # print(self.result)
            self.result.append(path[:]) # 深浅拷贝的问题
            return
        for i in range(startIndex, n + 1):
            path.append(i)
            # print(path)
            self.backTracking(i + 1, n, k, path)
            path.pop()
        return 
    def combine(self, n, k):
        """
        :type n: int
        :type k: int
        :rtype: List[List[int]]
        """
        path = []
        self.backTracking(1, n, k, path)
        return self.result
        
```

```python
# 17. 电话号码的字母组合
class Solution(object):
    def __init__(self):
        self.result = []
        self.map = {
            "1":"",
            "2":"abc",
            "3":"def",
            "4":"ghi",
            "5":"jkl",
            "6":"mno",
            "7":"pqrs",
            "8":"tuv",
            "9":"wxyz"
        }
    def backTacking(self, index, path, digits):
        if index == len(digits):
            self.result.append(path)
            return
        possibleLetters = self.map[digits[index]]
        possibleLetters = list(possibleLetters)
        for i in range(len(possibleLetters)):
            path += possibleLetters[i]
            self.backTacking(index + 1, path, digits)
            path = path[:-1]

    def letterCombinations(self, digits):
        """
        :type digits: str
        :rtype: List[str]
        """
        if len(digits) == 0:
            return []
        path = ""
        self.backTacking(0, path, digits)
        return self.result
        
```
