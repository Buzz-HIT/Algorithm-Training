```python
# 739.每日温度
class Solution(object):
    def dailyTemperatures(self, temperatures):
        """
        :type temperatures: List[int]
        :rtype: List[int]
        """
        n = len(temperatures)
        result = [0] * n
        singleStack = []
        for i in range(n):
            if len(singleStack) == 0 or temperatures[i] <= singleStack[-1][0]:
                singleStack.append((temperatures[i], i))
            else:
                while len(singleStack) != 0 and temperatures[i] > singleStack[-1][0]:
                    s = singleStack.pop()
                    result[s[1]] = i - s[1]
                singleStack.append((temperatures[i], i))
        return result
## 不用存两个元组，直接存下标也可以
        
```

```python
# 496.下一个更大元素 I
class Solution(object):
    def nextGreaterElement(self, nums1, nums2):
        """
        :type nums1: List[int]
        :type nums2: List[int]
        :rtype: List[int]
        """
        result = [-1]*len(nums1)
        stack = [0]
        for i in range(1,len(nums2)):
            # 情况一情况二
            if nums2[i]<=nums2[stack[-1]]:
                stack.append(i)
            # 情况三
            else:
                while len(stack)!=0 and nums2[i]>nums2[stack[-1]]:
                    if nums2[stack[-1]] in nums1:
                        index = nums1.index(nums2[stack[-1]])
                        result[index]=nums2[i]
                    stack.pop()                 
                stack.append(i)
        return result
        
```

```python
# 503.下一个更大元素 II
class Solution(object):
    def nextGreaterElements(self, nums):
        """
        :type nums: List[int]
        :rtype: List[int]
        """
        singleStack = []
        n = len(nums)
        result = [-1] * n
        for i in range(2 * n):
            if len(singleStack) == 0 or nums[i % n] <= nums[singleStack[-1] % n]:
                singleStack.append(i)
            else:
                while len(singleStack) != 0 and nums[i % n] > nums[singleStack[-1] % n]:
                    index = singleStack.pop() % n
                    result[index] = nums[i % n]
                singleStack.append(i)
        return result
```
