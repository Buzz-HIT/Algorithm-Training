```python
n, bag_weight = map(int, input().split())
weight = []
value = []
for _ in range(n):
    w, v = map(int, input().split())
    weight.append(w)
    value.append(v)

dp = [[0] * (bag_weight + 1) for _ in range(n)]
for i in range(0, bag_weight + 1):
    dp[0][i] = (i // weight[0]) * value[0]

for i in range(1, n):
    for j in range(1, bag_weight + 1):
        if j < weight[i]:
            dp[i][j] = dp[i - 1][j]
        else:
            dp[i][j] = max(dp[i - 1][j], dp[i][j - weight[i]] + value[i])
print(dp[n - 1][bag_weight])
```

```python
class Solution(object):
    def change(self, amount, coins):
        """
        :type amount: int
        :type coins: List[int]
        :rtype: int
        """
        dp = [[0] * (amount + 1) for _ in range(len(coins))]

        for i in range(0, len(coins)):
            dp[i][0] = 1
        for i in range(0, amount + 1):
            if i % coins[0] == 0:
                dp[0][i] = 1
        for i in range(1, len(coins)):
            for j in range(1, amount + 1):
                dp[i][j] = dp[i - 1][j]
                if j >= coins[i]:
                    dp[i][j] += dp[i][j - coins[i]]
        return dp[len(coins) - 1][amount]
```

```python
class Solution:
    def combinationSum4(self, nums, target):
        # dp[][j]和为j的组合的总数
        dp = [[0] * (target+1) for _ in nums]
        
        for i in range(len(nums)):
            dp[i][0] = 1
            
        # 这里不能初始化dp[0][j]。dp[0][j]的值依赖于dp[-1][j-nums[0]]
            
        for j in range(1, target+1):
            for i in range(len(nums)):
                
                if j - nums[i] >= 0:
                    dp[i][j] = (
                        # 不放nums[i]
                        # i = 0 时，dp[-1][j]恰好为0，所以没有特殊处理
                        dp[i-1][j] +
                        # 放nums[i]。对于和为j的组合，只有试过全部物品，才能知道有几种组合方式。所以取最后一个物品dp[-1][j-nums[i]]
                        dp[-1][j-nums[i]]
                    )
                else:
                    dp[i][j] = dp[i-1][j]
        return dp[-1][-1]
```
