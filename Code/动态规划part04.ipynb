```python
# 1049.最后一块石头的重量II
class Solution(object):
    def lastStoneWeightII(self, stones):
        """
        :type stones: List[int]
        :rtype: int
        """
        stoneSum = sum(stones)
        target = stoneSum // 2
        dp = [0] * stoneSum
        for i in stones:
            for j in range(target, i - 1, -1):
                dp[j] = max(dp[j], dp[j - i] + i)
        return abs(stoneSum - 2*dp[target])
        
```

```python
# 494.目标和
class Solution(object):
    def findTargetSumWays(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: int
        """
        numSum = sum(nums)
        if abs(target) > numSum:
            return 0
        if (target + numSum) % 2 == 1:
            return 0
        target_sum = (target + numSum) // 2
        dp = [[0] * (target_sum + 1) for _ in range(len(nums))]
        dp[0][0] = 1
        if nums[0] <= target_sum:
            dp[0][nums[0]] = 1
        numZero = 0
        for i in range(len(nums)):
            if nums[i] == 0:
                numZero += 1
            dp[i][0] = int(math.pow(2, numZero))
        for i in range(1, len(nums)):
            for j in range(1, target_sum + 1):
                dp[i][j] = dp[i - 1][j]
                if j >= nums[i]:
                    dp[i][j] += dp[i - 1][j - nums[i]]
        return dp[len(nums) - 1][target_sum]

        
```

```python
# 474.一和零
class Solution(object):
    def findMaxForm(self, strs, m, n):
        """
        :type strs: List[str]
        :type m: int
        :type n: int
        :rtype: int
        """
        dp = [[0] * (n + 1) for _ in range(m + 1)]
        print(dp)
        for s in strs:
            zeroNum = s.count('0')
            oneNum = len(s) - zeroNum
            for i in range(m, zeroNum - 1, -1):
                for j in range(n, oneNum - 1, -1):
                    dp[i][j] = max(dp[i][j], dp[i - zeroNum][j - oneNum] + 1)
        return dp[m][n]
```
