```python
## 491.递增子序列
class Solution(object):
    def __init__(self):
        self.result = []
    def backTracking(self, startIndex, path, nums):
        if len(path) > 1:
            self.append(path[:])
        uset = set()
        for i in range(startIndex, len(nums)):
            if (path and nums[i] < path[-1]) or nums[i] in uset:
                continue
            uset.add(nums[i])
            path.append(nums[i])
            self.backTracking(i + 1, path, nums)
            path.pop()
         
    def findSubsequences(self, nums):
        """
        :type nums: List[int]
        :rtype: List[List[int]]
        """
        self.backTracking(0, [], nums)
        return self.result
        
```

```python
## 46.全排列
class Solution(object):
    def __init__(self):
        self.result = []
    def backTracking(self, nums, used, path):
        if len(path) == len(nums):
            self.result.append(path[:])
            return
        for i in range(len(nums)):
            if used[i]:
                continue
            used[i] = 1
            path.append(nums[i])
            self.backTracking(nums, used, path)
            used[i] = 0
            path.pop()
    def permute(self, nums):
        """
        :type nums: List[int]
        :rtype: List[List[int]]
        """
        used = [0] * len(nums)
        self.backTracking(nums, used, [])
        return self.result
        
```

```python
## 47.全排列II
class Solution(object):
    def __init__(self):
        self.result = []
    def backTracking(self, nums, used, path):
        if len(path) == len(nums):
            self.result.append(path[:])
            return
        for i in range(len(nums)):
            if used[i] or (i > 0 and nums[i] == nums[i - 1]):
                continue
            used[i] = 1
            path.append(nums[i])
            self.backTracking(nums, used, path)
            used[i] = 0
            path.pop()
    def permute(self, nums):
        """
        :type nums: List[int]
        :rtype: List[List[int]]
        """
        used = [0] * len(nums)
        nums.sort()
        self.backTracking(nums, used, [])
        return self.result
        
```
