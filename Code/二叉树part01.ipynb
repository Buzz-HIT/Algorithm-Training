```python
# 144.二叉树的前序遍历（递归版）
# Definition for a binary tree node.
# class TreeNode(object):
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution(object):
    def __init__(self):
        self.ans = []
    def preorderTraversal(self, root):
        """
        :type root: Optional[TreeNode]
        :rtype: List[int]
        """
        self.preTravel(root)
        return self.ans
    def preTravel(self, cur):
        if cur == None:
            return
        self.ans.append(cur.val)
        self.preTravel(cur.left)
        self.preTravel(cur.right)
    
```

```python
# 144.二叉树的后序遍历（递归版）
# Definition for a binary tree node.
# class TreeNode(object):
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution(object):
    def __init__(self):
        self.ans = []
    def postorderTraversal(self, root):
        """
        :type root: Optional[TreeNode]
        :rtype: List[int]
        """
        self.postTravel(root)
        return self.ans
    def postTravel(self, cur):
        if cur == None:
            return
        self.postTravel(cur.left)
        self.postTravel(cur.right)
        self.ans.append(cur.val)
    
```

```python
# 144.二叉树的中序遍历（递归版）
# Definition for a binary tree node.
# class TreeNode(object):
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution(object):
    def __init__(self):
        self.ans = []
    def inorderTraversal(self, root):
        """
        :type root: Optional[TreeNode]
        :rtype: List[int]
        """
        self.inTravel(root)
        return self.ans
    def inTravel(self, cur):
        if cur == None:
            return
        self.inTravel(cur.left)
        self.ans.append(cur.val)
        self.inTravel(cur.right)
    
```

```python
# 统一风格的二叉树遍历
def preorderTraversal(self, root):
    result = []
    st = [(root, False)] if root else []
    while st:
        node, visited= st.pop()
        if visited:
            result.append(node.val)
            continue
        if node.right:
            st.append((node.right, False))
        st.append((node, True))
        if node.left:
            st.append((node.left))
    return result
```

```python
# 层次遍历空指针法
# Definition for a binary tree node.
# class TreeNode(object):
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution(object):
    def levelOrder(self, root):
        """
        :type root: Optional[TreeNode]
        :rtype: List[List[int]]
        """
        queue = [root, None] if root else [None]
        result = []
        layRes = []
        while len(queue) > 1:
            node = queue.pop(0)
            if node:
                layRes.append(node.val)
                if node.left:
                    queue.append(node.left)
                if node.right:
                    queue.append(node.right)
            else:
                queue.append(None)
                result.append(layRes)
                layRes = []
        if layRes:
            result.append(layRes)
        return result
```

```python
# 层次遍历官方解法
# Definition for a binary tree node.
# class TreeNode(object):
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution(object):
    def levelOrder(self, root):
        """
        :type root: Optional[TreeNode]
        :rtype: List[List[int]]
        """
        if not root:
            return []
        queue = collections.deque([root])
        result = []
        while queue:
            level = []
            for _ in range(len(queue)): # 这里就已经取到当前层的长度了，因此后面再加也就不会遍历了
                cur = queue.popleft()
                level.append(cur.val)
                if cur.left:
                    queue.append(cur.left)
                if cur.right:
                    queue.append(cur.right)
            result.append(level)
        return result
```

```python
# 二叉树的右视图
# Definition for a binary tree node.
# class TreeNode(object):
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution(object):
    def rightSideView(self, root):
        """
        :type root: Optional[TreeNode]
        :rtype: List[int]
        """
        if not root:
            return []
        result = []
        queue = collections.deque([root])
        while queue:
            l = len(queue)
            for _ in range(l):
                cur = queue.popleft()
                if cur.left:
                    queue.append(cur.left)
                if cur.right:
                    queue.append(cur.right)
                if _ == l - 1:
                    result.append(cur.val)
        return result
```
