```python
# 42.接雨水
class Solution(object):
    def trap(self, height):
        """
        :type height: List[int]
        :rtype: int
        """
        result = 0
        singleStack = []
        singleStack.append(0)
        for i in range(1, len(height)):
            if height[i] < height[singleStack[-1]]:
                singleStack.append(i)
            elif height[i] == height[singleStack[-1]]:
                singleStack.pop()
                singleStack.append(i)
            else:
                while singleStack and height[i] > height[singleStack[-1]]:
                    mid_height = height[singleStack[-1]]
                    singleStack.pop()
                    if singleStack:
                        rightHeight = height[i]
                        leftHeight = height[singleStack[-1]]
                        h = min(leftHeight, rightHeight) - mid_height
                        w = i - singleStack[-1] - 1
                        result += h * w
                singleStack.append(i)
        return result
        
```

```python
class Solution(object):
    def largestRectangleArea(self, heights):
        """
        :type heights: List[int]
        :rtype: int
        """
        res = 0
        heights.insert(0, 0)
        heights.append(0)
        stack = [0]
        result = 0
        for i in range(1, len(heights)):
            if heights[i] > heights[stack[-1]]:
                stack.append(i)
            elif heights[i] == heights[stack[-1]]:
                stack.pop()
                stack.append(i)
            else:
                while stack and heights[i] < heights[stack[-1]]:
                    mid_index = stack[-1]
                    stack.pop()
                    if stack:
                        left_index = stack[-1]
                        right_index = i
                        w = right_index - left_index - 1
                        h = heights[mid_index]
                        result = max(result, h * w)
                stack.append(i)
        return result
        
```
