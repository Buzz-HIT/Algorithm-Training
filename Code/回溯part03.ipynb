```python
# 93.复原IP地址
class Solution(object):
    def __init__(self):
        self.result = []
    def backTracking(self, s, startIndex, path):
        if startIndex == len(s) and len(path) == 4:
            self.result.append('.'.join(path))
            return
        if len(path) > 4:
            return
        for i in range(startIndex, len(s)):
            curStr = s[startIndex:i + 1]
            if int(curStr) >= 0 and int(curStr) <= 255:
                # curStr = str(int(curStr))
                if (len(curStr) > 1 and curStr[0] == '0'):
                    return
                path.append(curStr)
                self.backTracking(s, i + 1, path)
                path.pop()
    def restoreIpAddresses(self, s):
        """
        :type s: str
        :rtype: List[str]
        """
        self.backTracking(s, 0, [])
        return self.result
        
```

```python
# 78.子集
class Solution(object):
    def __init__(self):
        self.result = []
    def backTracking(self, starIndex, nums, path):
        self.result.append(path[:])
        for i in range(starIndex, len(nums)):
            path.append(nums[i])
            self.backTracking(i + 1, nums, path)
            path.pop()

    def subsets(self, nums):
        """
        :type nums: List[int]
        :rtype: List[List[int]]
        """
        self.backTracking(0, nums, [])
        return self.result        
```
