```python
# 509.斐波那契数
class Solution(object):
    def fib(self, n):
        """
        :type n: int
        :rtype: int
        """
        if n == 0:
            return 0
        if n == 1:
            return 1
        dp = [0] * (n + 1)
        dp[0] = 0
        dp[1] = 1
        for i in range(2, n + 1):
            dp[i] = dp[i - 1] + dp[i - 2]
        return dp[n]
```

```python
# 509.斐波那契数，空间复杂度O(1)
class Solution(object):
    def fib(self, n):
        """
        :type n: int
        :rtype: int
        """
        pre = 0
        cur = 1
        if n == 0:
            return pre
        if n == 1:
            return cur
        result = 0
        i = 2
        while i <= n:
            result = cur + pre
            pre = cur
            cur = result
            i += 1
        return result
```

```python
# 70.爬楼梯
class Solution(object):
    def climbStairs(self, n):
        """
        :type n: int
        :rtype: int
        """
        dp = [0] * (n + 1)
        if n == 0:
            return 0
        if n == 1:
            return 1
        if n == 2:
            return 2
        dp[0] = 0
        dp[1] = 1
        dp[2] = 2
        for i in range(3, n + 1):
            dp[i] = dp[i - 1] + dp[i - 2]
        return dp[n]
        
```

```python
#746.使用最小花费爬楼梯
class Solution(object):
    def minCostClimbingStairs(self, cost):
        """
        :type cost: List[int]
        :rtype: int
        """
        n = len(cost)
        dp = [0] * (n + 1)
        dp[0] = 0
        dp[1] = 0
        for i in range(2, n + 1):
            dp[i] = min((dp[i - 2] + cost[i - 2]), (dp[i - 1] + cost[i - 1]))
        return dp[n]
```
