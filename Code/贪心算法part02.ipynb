```python
# 122. 买卖股票的最佳时机 II
class Solution(object):
    def maxProfit(self, prices):
        """
        :type prices: List[int]
        :rtype: int
        """
        dayProfit = []
        for i in range(len(prices) - 1):
            dayProfit.append(prices[i + 1] - prices[i])
        totalProfit = 0
        for i in dayProfit:
            if i > 0:
                totalProfit += i
        return totalProfit
        
```

```python
# 55.跳跃游戏
class Solution(object):
    def canJump(self, nums):
        """
        :type nums: List[int]
        :rtype: bool
        """
        cover = 0
        i = 0
        while i <= cover:
            cover = max(i + nums[i], cover)
            # print(cover)
            if cover >= len(nums) - 1: # cover就是当前覆盖的最大区域，每次循环就在当前可覆盖的最大区域中去找下面的最大区域
                return True
            i += 1
        return False
        
```

```python
# 45.跳跃游戏 II
class Solution(object):
    def jump(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        if len(nums) == 1:
            return 0
        curDistance = 0
        nextDistance = 0
        ans = 0
        for i in range(len(nums)):
            nextDistance = max(i + nums[i], nextDistance)
            if i == curDistance:
                ans += 1
                curDistance = nextDistance # 这个时候遍历完了当前的范围，所以nexr一定取得是最大值
                if (nextDistance >= len(nums) - 1):
                    break
        return ans

        
```

```python
# 1005. K 次取反后最大化的数组和
class Solution(object):
    def largestSumAfterKNegations(self, nums, k):
        """
        :type nums: List[int]
        :type k: int
        :rtype: int
        """
        nums.sort(key = lambda x : abs(x), reverse=True)
        for i in range(len(nums)):
            if nums[i] < 0 and k > 0:
                #print(nums)
                nums[i] *= -1
                #print(nums)
                k -= 1
        if k % 2 == 1: # k为偶数，不用管，反复旋转最小的即可
            nums[-1] *= -1
        return sum(nums)


```
