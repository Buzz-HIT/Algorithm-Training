```python
# 110.平衡二叉树
# Definition for a binary tree node.
# class TreeNode(object):
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution(object):
    def getHeight(self, root):
        if root==None:
            return 0
        leftHeight = self.getHeight(root.left)
        if leftHeight == -1:
            return -1
        rightHeight = self.getHeight(root.right)
        if rightHeight == -1:
            return -1
        if abs(leftHeight - rightHeight) > 1:
            return -1
        return 1 + max(leftHeight, rightHeight)
    def isBalanced(self, root):
        """
        :type root: Optional[TreeNode]
        :rtype: bool
        """
        return self.getHeight(root) != -1
        
```

```python
# 257. 二叉树的所有路径
# Definition for a binary tree node.
# class TreeNode(object):
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution(object):
    def __init__(self):
        self.result = []
    def searchPath(self, root, path):
        path.append(root.val)
        if root.left == None and root.right == None:
            self.result.append('->'.join(map(str, path))) # 注意这里，path算是公共变量，在每个递归栈中共享空间
            return
        if root.left:
            self.searchPath(root.left, path)
            path.pop()
        if root.right:
            self.searchPath(root.right, path)
            path.pop()
        return

    def binaryTreePaths(self, root):
        """
        :type root: Optional[TreeNode]
        :rtype: List[str]
        """
        if root == None:
            return self.result
        path = []
        self.searchPath(root, path)
        return self.result
        
```

```python
# 404. 左叶子之和
# Definition for a binary tree node.
# class TreeNode(object):
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution(object):
    def __init__(self):
        self.sum = 0
    def travelTree(self, root):
        if root == None:
            return
        if root.left != None and (root.left.left == None and root.left.right == None):
            self.sum += root.left.val
            self.travelTree(root.right)
            return
        else:
            self.travelTree(root.left)
            self.travelTree(root.right)
            return 
    def sumOfLeftLeaves(self, root):
        """
        :type root: Optional[TreeNode]
        :rtype: int
        """
        if root == None:
            return 0
        self.travelTree(root)
        return self.sum

            

        
```

```python
# 222.完全二叉树的节点个数（标准）
class Solution(object):
    def getNodeNum(self, cur):
        if cur == None:
            return 0
        leftNum = self.getNodeNum(cur.left)
        rightNum = self.getNodeNum(cur.right)
        return leftNum + rightNum + 1
    def countNodes(self, root):
        """
        :type root: Optional[TreeNode]
        :rtype: int
        """
        return self.getNodeNum(root)
   
```

```python
# 完全二叉树的性质版本
# Definition for a binary tree node.
# class TreeNode(object):
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution(object):
    def countNodes(self, root):
        """
        :type root: Optional[TreeNode]
        :rtype: int
        """
        if root == None:
            return 0
        left = root.left
        right = root.right
        leftDepth = 0
        rightDepth = 0
        while left:
            leftDepth += 1
            left = left.left
        while right:
            rightDepth += 1
            right = right.right
        if leftDepth == rightDepth:
            # print((2 ** (leftDepth + 1)) - 1)
            return ((2 ** (leftDepth + 1)) - 1)
        return self.countNodes(root.left) + self.countNodes(root.right) + 1
```
