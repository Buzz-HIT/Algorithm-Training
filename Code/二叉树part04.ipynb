```python
#513.找树左下角的值
# Definition for a binary tree node.
# class TreeNode(object):
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution(object):
    def findBottomLeftValue(self, root):
        """
        :type root: Optional[TreeNode]
        :rtype: int
        """
        queue = collections.deque()
        if root == None:
            return 0
        queue.append(root)
        leftNode = None
        while queue:
            size = len(queue)
            for _ in range(size):
                cur = queue.popleft()
                if _ == 0:
                    leftNode = cur
                if cur.left != None:
                    queue.append(cur.left)
                if cur.right != None:
                    queue.append(cur.right)
        return leftNode.val
                


        
```

```python
# 112. 路径总和
# Definition for a binary tree node.
# class TreeNode(object):
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution(object):
    def travelTree(self, root, count):
        if root.left == None and root.right == None:
            if count == 0:
                return True 
            else:
                return False
        if root.left:
            if self.travelTree(root.left, count - root.left.val): # 这种精简的方法，是在递归调用时直接返回结果，所以应该减的应该是其左节点的值
                return True
        if root.right:
            if self.travelTree(root.right, count - root.right.val):
                return True
        return False

    def hasPathSum(self, root, targetSum):
        """
        :type root: Optional[TreeNode]
        :type targetSum: int
        :rtype: bool
        """
        if root == None:
            return False
        return self.travelTree(root, targetSum - root.val)
        
```

```python
# 113. 路径总和II
# Definition for a binary tree node.
# class TreeNode(object):
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution(object):
    def __init__(self):
        self.result = []
    def travelTree(self, cur, count, path):
        if cur.left == None and cur.right == None:
            if count == 0:
                self.result.append(path[:]) # 这里要注意是path的拷贝，而不是path本身
            return
        if cur.left:
            path.append(cur.left.val)
            self.travelTree(cur.left, count - cur.left.val, path)
            path.pop()
        if cur.right:
            path.append(cur.right.val)
            self.travelTree(cur.right, count - cur.right.val, path)
            path.pop()
    def pathSum(self, root, targetSum):
        """
        :type root: Optional[TreeNode]
        :type targetSum: int
        :rtype: List[List[int]]
        """
        if root == None:
            return self.result
        path = []
        self.result = []
        path.append(root.val)
        self.travelTree(root, targetSum - root.val, path)
        return self.result
```

```python
# 106. 从中序与后序遍历序列构造二叉树
# Definition for a binary tree node.
# class TreeNode(object):
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution(object):
    def buildTree(self, inorder, postorder):
        """
        :type inorder: List[int]
        :type postorder: List[int]
        :rtype: Optional[TreeNode]
        """
        if len(postorder) == 0:
            return None
        val = postorder[-1]
        cur = TreeNode(val)
        i = 0
        while inorder[i] != val:
            i += 1
        inorderLeft = inorder[0:i]
        inorderRight = inorder[i+1:]
        postorderLeft = postorder[0:len(inorderLeft)]
        postorderRight = postorder[i:len(postorder) - 1]
        # print(inorderLeft, postorderLeft, inorderRight, postorderRight)
        cur.left = self.buildTree(inorderLeft, postorderLeft)
        cur.right = self.buildTree(inorderRight, postorderRight)
        return cur
        
```

```python
# 105. 从前序与中序遍历序列构造二叉树
# Definition for a binary tree node.
# class TreeNode(object):
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution(object):
    def buildTree(self, preorder, inorder):
        """
        :type preorder: List[int]
        :type inorder: List[int]
        :rtype: Optional[TreeNode]
        """
        if len(preorder) == 0:
            return None
        val = preorder[0]
        cur = TreeNode(val)
        i = 0
        while inorder[i] != val:
            i += 1
        inorderLeft = inorder[0:i]
        inorderRight = inorder[i+1:]
        preorderLeft = preorder[1:1+len(inorderLeft)]
        preorderRight = preorder[1+len(inorderLeft):]
        # print(inorderLeft, preorderLeft, inorderRight, preorderRight)
        cur.left = self.buildTree(preorderLeft, inorderLeft)
        cur.right = self.buildTree(preorderRight, inorderRight)
        return cur
        
```
